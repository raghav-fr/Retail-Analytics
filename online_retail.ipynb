```python
regex={"negative value":r'^-',"start with letter":r'^[A-Za-z]',"start with number":r'^[0-9]',"only letters":r'^[^\d]+$',"only numbers":r'^[0-9]+$',"end with letter":r'[A-Za-z]$',"any character in between 2 number":r'(?<=\d)[^\d](?=\d)',"any number in between two character":r'(?<=\[A-Za-z])[\d](?=\[A-Za-z])',"more than 1 characters present": r'[a-zA-Z].*[a-zA-Z]'}
count={}
for key,value in regex.items():
    count[key]=df['InvoiceNo'].astype(str).str.match(value).sum().item()
    
invCount = pd.DataFrame(list(count.items()), columns=["Pattern", "MatchCount"])
invCount
```

```python
matching=["na","n/a","not available","null","none"]
count={}
for i in matching:
    count[i]=df['InvoiceNo'].astype(str).str.contains(i).sum().item()
print(count)
```
